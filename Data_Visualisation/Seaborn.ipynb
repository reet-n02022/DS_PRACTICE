```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


var=[1,2,3,4,5,6,7]
var_1=[2,3,4,1,5,6,7]

x_1=pd.DataFrame({"var":var,"var_1":var_1})

sns.lineplot(x="var",y="var_1",data=x_1)
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=y_1,hue="sex",size=10,style="sex",
            palette="Accent",markers=["o",">"],dashes=True,legend="full")
plt.title("wscube")
plt.grid()
plt.show()
```

barplot

```python
 sns.barplot(x=y_1.island,y=y_1.bill_length_mm,color="g")
plt.show()
```

```python

order_1=["Dream","Torgersen","Biscoe"]
sns.barplot(x="island",y="bill_length_mm",data=y_1,hue="sex",order=order_1,hue_order=["Female","Male"],
           errorbar=("ci",20),orient="v",palette="icefire",saturation=10,err_kws={'color': 'y'},capsize=1,
           dodge=True,alpha=0.9)
sns.set(style="darkgrid")

plt.show()
```

```python

sns.barplot(x="bill_length_mm",y="bill_depth_mm",data=y_1,hue="sex",hue_order=["Female","Male"],
           errorbar=("ci",20),orient="h")
plt.show()
```

Histogram plot

```python
sns.displot(y_1["flipper_length_mm"],bins=[170,180,190,200,210,220,230,240],kde=True,rug=True,color="g",log_scale=False)
plt.show()
```

Scatter plot

```python
m={"Male":"*","Female":"o"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=y_1,hue="sex",style="sex",size="sex",sizes=(100,60),
               palette="Accent",alpha=1,markers=m)
plt.show()
```

Heatmap

```python
var_2=np.linspace(1,10,20).reshape(4,5)
var_2
```

```python
y={"fontsize":20,"color":"r"}
v=sns.heatmap(var_2,vmin=0,vmax=10,cmap="PuOr",annot=True, annot_kws=y,linewidth=10,linecolor="y",cbar=True,
           xticklabels=False,yticklabels=False)
v.set(xlabel="python",ylabel="wscube")
sns.set(font_scale=1)
plt.show()
```

Count plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data1=sns.load_dataset("tips")
data1
```

```python
sns.countplot(y="sex",data=data1,hue="smoker",palette="bwr",saturation=0.7)
plt.show()

```

```python
sns.barplot(x="sex",y="size",data=data1)
plt.show()
```

Violin Plot

```python
sns.violinplot(x="day",y="total_bill",data=data1, hue="time",palette="Dark2",saturation=0.3)
plt.show()
```

```python
sns.violinplot(x="time",y="total_bill",data=data1,linewidth=2,order=["dinner","Lunch"],inner="point") #{"quartile","point","stick",None}
plt.show()
```

```python
sns.violinplot(y="day",x="total_bill",data=data1,hue="sex",split=True,density_norm="width")
plt.show()
```

```python
sns.violinplot(y=data1["total_bill"])
plt.show()
```

Pair Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data2=sns.load_dataset("tips")
data2
```

```python
sns.pairplot(data2,hue="sex",kind="reg") #"scatter","kde","hist","reg"
plt.show()
```

```python
sns.pairplot(data2, vars=["total_bill","tip"],hue="sex",hue_order=["Female","Male"],palette="BuGn",)
plt.show()
```

```python
sns.pairplot(data2,hue="sex",hue_order=["Female","Male"],palette="BuGn",x_vars=["total_bill","tip"])
plt.show()
```
